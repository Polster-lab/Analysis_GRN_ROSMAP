```python

def return_gene_name(gene):
    
    try:
        if ensemble_to_gene_name[gene] == 'Gene name not found':
            return gene
        return ensemble_to_gene_name[gene]
    except:
        return gene
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from tqdm import tqdm

dct_corr = dict()
dct_p_value = dict()
s = 0
for i in all_clinical_data_numerical_columns[clinical_var].columns:
    
    df = all_clinical_data_numerical_columns[i].dropna()
    df = pd.DataFrame(df)
    
    temp_data = data.loc[df.index]
    clinical_var_list = df[i].tolist()
    
    for k in l:
        edge_weights_list = temp_data[k].tolist()

        
        
        # Calculate Spearman correlation
        rho, p_value = stats.spearmanr(clinical_var_list, edge_weights_list)
    
        dct_corr[f'{i}_{k}'] = rho
        dct_p_value[f'{i}_{k}'] = p_value
        if p_value < 0.05:
            s = s + 1
        # Plot correlation
            plt.figure(figsize=(6, 4))
            sns.regplot(x=clinical_var_list, y=edge_weights_list, scatter_kws={"s": 50}, line_kws={"color": "red"})
            tf_name = return_gene_name(k.split('_')[0])
            gene_name = return_gene_name(k.split('_')[1])
            
            # Add title and labels
            plt.title(f'Correlation: {i} vs {tf_name}-{gene_name}')
            plt.xlabel(i)
            plt.ylabel(f'{tf_name}-{gene_name}')
            
            # Annotate correlation values
            plt.text(min(clinical_var_list), max(edge_weights_list)-1, 
                     f"ρ = {rho:.2f}\nP-value = {p_value:.3f}", 
                     fontsize=9, bbox=dict(facecolor='white', alpha=0.5))
    
            # Show plot
            plt.show()
           
```

```python
dct = {
      
    'cognse_demog_slope': 'β Semantic Memory ~ Demographics', 
    'cognep_demog_slope': 'β Episodic Memory ~ Demographics', 
    'cognwo_path_slope': 'β  Working Memory ~ Demographic + Pathology', 
    'cogng_demog_slope': 'β Global Cognition ~ Demographics', 
    'cognpo_demog_slope': 'β Visuospatial ability ~ Demographics', 
    'cogng_path_slope': 'β Global Cognition ~ Demographics + Pathology', 
    'cognwo_demog_slope': 'β Working Memory ~ Demographics', 
    'cognep_path_slope': 'β Episodic Memory ~ Demographics + Pathology',  
}
```

```python
def return_cog_name(x):
    return dct[x]
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

dct_corr = dict()
dct_p_value = dict()

# List to store significant plots
significant_plots = []

for i in all_clinical_data_numerical_columns[clinical_var].columns:
    
    df = all_clinical_data_numerical_columns[i].dropna()
    df = pd.DataFrame(df)
    
    temp_data = data.loc[df.index]
    clinical_var_list = df[i].tolist()
    
    for k in l:
        edge_weights_list = temp_data[k].tolist()

        # Calculate Spearman correlation
        rho, p_value = stats.spearmanr(clinical_var_list, edge_weights_list)
        
        dct_corr[f'{i}_{k}'] = rho
        dct_p_value[f'{i}_{k}'] = p_value
        
        # Store only significant correlations
        if p_value < 0.05:
            significant_plots.append((i, k, rho, p_value, clinical_var_list, edge_weights_list))

# Set up subplots grid (3 rows, 4 columns)
num_plots = len(significant_plots)
rows = (num_plots // 3) + (1 if num_plots % 3 else 0)  # Calculate needed rows
cols = 3  # Always 4 columns

fig, axes = plt.subplots(rows, cols, figsize=(16, 4 * rows))  # Dynamic row count
axes = axes.flatten()  # Flatten in case of 1 row

# Plot each significant correlation
for idx, (i, k, rho, p_value, clinical_var_list, edge_weights_list) in enumerate(significant_plots):
    
    tf_name = return_gene_name(k.split('_')[0])
    gene_name = return_gene_name(k.split('_')[1])
    
    ax = axes[idx]
    sns.regplot(x=clinical_var_list, y=edge_weights_list, scatter_kws={"s": 50}, line_kws={"color": "red"}, ax=ax)
    y_axis_name = return_cog_name(i)
    # Add title and labels
    ax.set_title(f'Correlation: {y_axis_name} vs {tf_name}-{gene_name}', fontsize = 9)
    ax.set_xlabel(f'{y_axis_name}')
    ax.set_ylabel(f'{tf_name}-{gene_name}')
    
    # Annotate correlation values
    ax.text(min(clinical_var_list), max(edge_weights_list) - 1, 
            f"ρ = {rho:.2f}\nP-value = {p_value:.3f}", 
            fontsize=9, bbox=dict(facecolor='white', alpha=0.5))

# Hide unused subplots
for idx in range(len(significant_plots), len(axes)):
    fig.delaxes(axes[idx])

plt.tight_layout() 
plt.savefig("./../../../data_preprocessed/grn_analysis_output/plots/correlation_plots.svg", format="svg", dpi=600)# Adjust layout
plt.show()
```

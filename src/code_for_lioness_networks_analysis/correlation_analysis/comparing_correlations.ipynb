```python
gene_sets = ['GO_Molecular_Function_2023','GO_Cellular_Component_2023','GO_Biological_Process_2023','KEGG_2021_Human','Reactome_Pathways_2024','Reactome_2022']



def return_gseapy_obj(gene_rank, gene_set):
    
    pre_res = gp.prerank(rnk=gene_rank, # or rnk = rnk,
                     gene_sets=gene_set,
                     threads=4,
                     min_size=5,
                     max_size=1000,
                     permutation_num=1000, # reduce number to speed up testing
                     outdir=None, # don't write to disk
                     seed=6,
                     verbose=True, # see what's going on behind the scenes
                    )
    return pre_res
    
```

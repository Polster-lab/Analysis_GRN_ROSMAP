```python
import numpy as np
import pandas as pd
from scipy.stats import mannwhitneyu
import seaborn as sns
import matplotlib.pyplot as plt
p_values = []
for i in range(5):  # Modify the range according to the number of columns
    ad_data = ad_metrics_df.iloc[:, i]
    nci_data = nci_metrics_df.iloc[:, i]
    
    # Perform Mann-Whitney U test to compare AD and NCI
    stat, p_value = mannwhitneyu(ad_data, nci_data, alternative='two-sided')
    
    # Store the p-value
    p_values.append(p_value)

# Calculate the median p-value
median_p_value = np.median(p_values)
```

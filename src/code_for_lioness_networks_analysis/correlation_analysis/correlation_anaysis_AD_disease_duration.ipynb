```python
from scipy import stats

dct_corr= dict()
dct_p_value= dict()
from tqdm import tqdm
for i in tqdm(all_clinical_data_numerical_columns.columns):
    
    df = all_clinical_data_numerical_columns[i].dropna()
    
    df = pd.DataFrame(df)
    
    temp_data = data.loc[df.index]

    clinical_var_list = df[i].tolist()
    
    #print(temp_data)
    for k in temp_data.columns:
        
        #print(clinical_var_list)
        
        #print(temp_df_count_matrix)
        edge_weights_list = temp_data[k].tolist()
        
        rho, p_value = stats.spearmanr(clinical_var_list,edge_weights_list)

        dct_corr[f'{i}_{k}']= rho
        
        dct_p_value[f'{i}_{k}'] = p_value 
```

```python
corr_df = pd.DataFrame.from_dict(list(dct_corr.items()))
```

```python
corr_df = corr_df.sort_values(by = 1)
corr_df
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report, f1_score
from sklearn.model_selection import cross_val_predict, cross_val_score

# Assuming your DataFrame contains only numeric values
# If not, you might need to preprocess your data accordingly

# Split the data into features (X) and target (y)
X = outlier_corrected_combined.drop(columns='Label')[l]
y = outlier_corrected_combined['Label']

# Initialize the model
model = LogisticRegression(class_weight={'AD': 1, 'NCI': 1})

# 5-Fold Cross-Validation
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Store confusion matrices and F1 scores
conf_matrices = []
f1_scores = []

# Get cross-validation predictions and evaluate accuracy for each fold
accuracies = cross_val_score(model, X, y, cv=cv, scoring='accuracy')
y_pred = cross_val_predict(model, X, y, cv=cv)

# Print the average accuracy
print(f"Average Accuracy: {accuracies.mean():.2f} ± {accuracies.std():.2f}")

# Create boxplot of accuracies across folds
plt.figure(figsize=(10, 6))
sns.boxplot(data=accuracies)
plt.title('Cross-Validation Accuracy Distribution')
plt.ylabel('Accuracy')
plt.show()

# Compute confusion matrix and F1 score for each fold
for train_index, test_index in cv.split(X, y):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
    # Fit the model and predict for this fold
    model.fit(X_train, y_train)
    y_fold_pred = model.predict(X_test)
    
    # Compute confusion matrix for this fold
    fold_conf_matrix = confusion_matrix(y_test, y_fold_pred)
    conf_matrices.append(fold_conf_matrix)
    
    # Compute F1 score for this fold
    f1 = f1_score(y_test, y_fold_pred, average='weighted')  # weighted F1 score for imbalanced classes
    f1_scores.append(f1)

# Plot confusion matrix for each fold
fig, axes = plt.subplots(1, 5, figsize=(20, 4))  # 5 subplots for 5 folds
for i, (conf_matrix, ax) in enumerate(zip(conf_matrices, axes)):
    sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=y.unique(), yticklabels=y.unique(), cbar=False)
    ax.set_title(f'Fold {i+1}')
    ax.set_xlabel('Predicted')
    ax.set_ylabel('True')

plt.tight_layout()
plt.savefig("./../../data_preprocessed/grn_analysis_output/plots/5_fold_confusion_matrix.svg", format="svg", dpi=600)
plt.show()

# Print F1 scores for each fold
print("\nF1 Scores for each fold:")
for i, f1 in enumerate(f1_scores, 1):
    print(f"Fold {i}: F1 Score = {f1:.2f}")

# Classification report
class_report = classification_report(y, y_pred)
print("\nClassification Report:")
print(class_report)
```

```python
import numpy as np

# Bootstrap method to compute p-value
def bootstrap_p_value(accuracies, given_accuracy, n_iter=10000):
    resampled_accuracies = np.random.choice(accuracies, (n_iter, len(accuracies)), replace=True)
    resampled_means = resampled_accuracies.mean(axis=1)
    p_value = np.mean(resampled_means >= given_accuracy)
    return p_value

# Calculate bootstrap p-value
p_value_bootstrap = bootstrap_p_value(accuracies, given_accuracy)
print(f"Bootstrap p-value: {p_value_bootstrap:.4f}")
```

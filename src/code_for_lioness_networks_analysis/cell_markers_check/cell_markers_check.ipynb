```python
import pickle

# Path to your pickle file
pickle_file_path = '../../../data_preprocessed/pickle_file/ensemble_to_gene_name.pickle'

# Load the pickle file
with open(pickle_file_path, 'rb') as file:
    ensemble_to_gene_name = pickle.load(file)



```

```python
def return_gene_name(gene):
    
    try:
        if ensemble_to_gene_name[gene] == 'Gene name not found':
            return gene
        return ensemble_to_gene_name[gene]
    except:
        return gene
```

```python
important_tf_genes = [return_gene_name(i) for i in important_tf_genes]
```

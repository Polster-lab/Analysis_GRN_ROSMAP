```python
from netZooPy.panda import Panda
from netZooPy.lioness import Lioness
from netZooPy.lioness.analyze_lioness import AnalyzeLioness
import pandas as pd
```

# NCI

```python
exp_data_path = '../../data_preprocessed/grn_input/count_matrix_nci.txt'
motif_data_path = '../../data_preprocessed/grn_input/tf.txt' 
ppi_data_path = '../../data_preprocessed/grn_input/ppi.txt'
```

```python
ppi_data = pd.read_csv(ppi_data_path ,header=None, sep = "\t")
ppi_data
```

```python
motif_data = pd.read_csv(motif_data_path ,header=None, sep = "\t")
motif_data
```

```python
expression_data = pd.read_csv(exp_data_path ,index_col=0, sep = "\t")
expression_data
```

# Panda

```python
panda_obj = Panda(exp_data_path,motif_data_path,ppi_data_path,
                  with_header = True,
                  remove_missing=False, 
                  keep_expression_matrix=True, save_memory=False, modeProcess='legacy')
```

```python
panda_output = '/12tb_dsk2/danish/GRN_new_21_feb_2025/data_preprocessed/grn_output/ad_nci_seperate_network/panda/output_panda_nci.txt'
```

```python
panda_obj.save_panda_results(panda_output)
```

```python
panda_obj.top_network_plot(top=10)
```

```python
panda_result_df = panda_obj.export_panda_results
```

```python
panda_result_df = panda_result_df.sort_values(by = 'force', ascending=False)
```

```python
panda_result_df
```

```python
panda_result_df.to_csv('/12tb_dsk2/danish/GRN_new_21_feb_2025/data_preprocessed/grn_output/ad_nci_seperate_network/panda/output_panda_nci_csv.csv',index = False)
```

```python
from tqdm import tqdm
col_1 = []
col_2 = []
values = []
for i in tqdm(os.listdir(raw_single_sample_network_dir_path)):
    if i == 'panda_obj.pkl' or i == 'output_panda_df.csv':
        continue
    data = pd.read_csv(raw_single_sample_network_dir_path + i, index_col=0)
    
    for i in tqdm(data.iterrows()):
        col_1.extend([i[0] for l in range(len(data.columns))])
        col_2.extend(data.columns)
        values.extend(i[1].tolist())
    preprocessed_data = pd.DataFrame([col_1,col_2,values]).T
    break
```

```python
for m in tqdm(os.listdir(raw_single_sample_network_dir_path)):
    col_1 = []
    col_2 = []
    values = []
    if m in os.listdir(preprocessed_single_sample_network_dir_path):
        continue
        
    
    if m == 'panda_obj.pkl' or m == 'output_panda_df.csv':
        continue
    if m == 'output_panda_df.csv':
        continue
    print(m)
    if 'csv' in m:
        data = pd.read_csv(raw_single_sample_network_dir_path + m, index_col=0)
        list_of_genes = data.columns
        list_of_tf = data.index.tolist()
        for i in list_of_tf:
            for j in list_of_genes:
                if data.loc[i,j] >= 3.69446:
                    col_1.append(i)
                    col_2.append(j)
                    values.append(data.loc[i,j])
                 
        preprocessed_data = pd.DataFrame([col_1, col_2, values]).T
        preprocessed_data.to_csv(preprocessed_single_sample_network_dir_path  + m, index= False)
```
